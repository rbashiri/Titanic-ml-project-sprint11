```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 
import seaborn as sns

data = pd.read_csv("/home/susan/Titanic-ml-project-sprint11/training/data/titanic.csv")

df = data.copy()
df.info()
```

```python
print(df.shape)
print(df.head(5))
```

```python

# Make column names lowercase and remove extra spaces
df.columns = df.columns.str.strip().str.lower()

# Convert passengerid to numeric
df['passengerid'] = pd.to_numeric(df['passengerid'], errors='coerce')
print(df['passengerid'].dtype)

# Check how many missing values are in each column
print(df.isna().sum())

# Show rows where age or cabin is missing
print(df[df[['age', 'cabin']].isna().any(axis=1)].head())

# Fill missing age values with the median age
df ['age']= df['age'].fillna(df['age'].median())
df['embarked']= df['embarked'].fillna(df['embarked'].mode()[0])
# Embarked means the port where the passenger got on the Titanic.
print(df['embarked'].mode()[0])
# Why use mode()[0]?
# mode() returns the most frequent value(s) in the column.
#[0] selects the first most common value.


# check for duplicated data
print(df.duplicated().sum())
# cabin has many missing values, so keep that information
df['cabin_missing'] = df['cabin'].isna().astype(int)

# fill missing cabin with a label
df['cabin'] = df['cabin'].fillna('Unknown')
df.info() 
```
